## 2. Build the Working Dataframe

We walk the extracted folders, keep only classes containing "Tomato", and record each image's filepath and label.

```
import os
import pandas as pd

base_dir = "plant_data/PlantVillage"
records = []

for class_name in os.listdir(base_dir):
    if "Tomato" in class_name:
        class_path = os.path.join(base_dir, class_name)
        for fname in os.listdir(class_path):
            records.append({
                "filepath": os.path.join(class_path, fname),
                "label": class_name
            })

df = pd.DataFrame(records)
df.head()
```

## 3. Expected Output Variable

For this dataset, the **expected output (target) variable** for the final application is the **`label`** column — the tomato disease class (e.g. Early Blight, Late Blight, Leaf Mold, Bacterial Spot, Healthy, etc.). The model will take a leaf image as input and predict this label.

```
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

fig, axes = plt.subplots(2, 4, figsize=(14, 7))
sample_rows = df.sample(8, random_state=1).reset_index(drop=True)

for i, ax in enumerate(axes.flat):
    img = mpimg.imread(sample_rows.loc[i, 'filepath'])
    ax.imshow(img)
    ax.set_title(sample_rows.loc[i, 'label'], fontsize=8)
    ax.axis('off')

plt.tight_layout()
plt.show()
```
